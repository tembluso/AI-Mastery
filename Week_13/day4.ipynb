Why this matters

Until now you trained on tiny toy datasets. But in practice, we use pretrained LLMs. Hugging Face gives you GPT-2, BERT, and hundreds of others ready for inference with just a few lines of code.

Theory Essentials

Hugging Face Transformers = hub for pretrained models.

Pipeline API makes inference easy (text-generation, fill-mask, sentiment).

GPT-2 → autoregressive generation (continue text).

BERT → fill-mask prediction (recover missing words).

Loading pretrained = avoids huge compute cost.

```python
# Setup
from transformers import pipeline

# GPT-2: text generation
gen = pipeline("text-generation", model="gpt2")
print("GPT-2:\n", gen("The future of AI is", max_length=30, num_return_sequences=1)[0]["generated_text"])

# BERT: fill-mask
fill = pipeline("fill-mask", model="bert-base-uncased")
print("BERT:\n", fill("The capital of France is [MASK].")[0])
```

Reflection

Why is it important that pretrained models exist for GPT/BERT?

What risks come from using models “as-is” without fine-tuning?

1. Why is it important that pretrained models exist for GPT/BERT?
Because pretraining on massive text corpora gives models broad language knowledge (grammar, vocabulary, world facts). Fine-tuning then only needs small task-specific data, making advanced NLP possible for everyone without billions of dollars in compute.

2. What risks come from using models “as-is” without fine-tuning?
They may give generic or biased outputs, fail at your specific domain (e.g., medical, legal text), or produce misleading results. Without fine-tuning, the model doesn’t adapt to your context, so accuracy, reliability, and safety can be poor.
