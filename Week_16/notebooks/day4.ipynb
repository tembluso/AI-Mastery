Week 16 · Day 4 — Frontend: Streamlit Chat UI
Why this matters

Today, you’ll turn your backend into something people can actually use.
You’ll build a simple Streamlit interface that lets users ask questions, displays answers, and shows the retrieved sources — your first interactive AI app.

Theory Essentials

Frontend ≠ Logic — it just calls your backend functions.

Streamlit turns Python functions into interactive UI elements.

Use st.chat_input() for questions and st.chat_message() for chat bubbles.

The frontend should display both answers and citations (source docs).

Keep UI state (chat history) with st.session_state.

Clean layout: title, input box, message display, and optional sidebar for settings.

code is in frontend.py

Mini-Challenge (≤40 min)

Goal: Build a clean, responsive chat interface that uses your backend in real time.

Acceptance Criteria
✅ Streamlit app loads without error (streamlit run app/frontend.py)
✅ Users can input a question and get an answer
✅ Sources are shown (in expander or below message)
✅ Persistent chat history across turns (st.session_state)

Notes / Key Takeaways

Streamlit allows fast iteration — no need for HTML/JS.

Keep backend logic out of the UI file.

Showing sources increases trust & transparency.

Save chat state to avoid losing conversation on refresh.
