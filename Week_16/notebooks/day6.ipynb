Week 16 · Day 6 — Evaluation & Hallucination Testing
Why this matters

Now that your Q&A bot is functional and transparent, it’s time to measure how well it performs.
Today you’ll evaluate how accurate the answers are and how often the model hallucinates — i.e., gives confident but unsupported answers.

Theory Essentials

Evaluation = Retrieval + Generation quality.

Use a small Q&A dataset (manual or public) to benchmark.

Precision@k / Recall@k: how often retrieved docs contain the correct answer.

Answer Accuracy: compare generated answers to ground truth.

Hallucination Rate: % of answers not supported by retrieved text.

You can start manually (small dataset), then automate with metrics later.

Acceptance Criteria
✅ Loads Q&A dataset (CSV or inline list).
✅ Computes Precision@k and Hallucination Rate.
✅ Displays a short table of questions, expected phrases, and hallucination flags.
✅ Prints summary report (e.g., “Precision@3 = 0.67, Hallucination = 0.33”).

Notes / Key Takeaways

Retrieval metrics (Precision@k) evaluate document relevance.

Hallucination checks evaluate faithfulness to sources.

Small manual datasets are fine — they teach you how to measure.

If hallucination rate is high → improve chunking, embeddings, or context length.

Reflection

Which metric (retrieval or generation) seems your current bottleneck?

How could evaluation logs guide retraining or better chunking?

1) Which metric (retrieval or generation) seems your current bottleneck?

Right now, the bottleneck is usually retrieval, because if the system pulls irrelevant or incomplete chunks, the LLM can’t fix it — it will either hallucinate or answer poorly.

Generation quality only matters once retrieval is solid; with only a couple of chunks in the index, precision@k and recall@k are the limiting factors.

2) How could evaluation logs guide retraining or better chunking?

Logs of queries + retrieved chunks + user feedback show when the system fails.

If many queries return irrelevant chunks → need better cleaning, smaller chunk size, or denser embeddings.

If answers cut off context → adjust overlap size to preserve continuity.

Over time, these logs form a dataset for retraining embeddings (e.g., fine-tuning on your domain) or for re-indexing with smarter chunking strategies.
